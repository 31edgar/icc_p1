# Prueba de gausspiv()
Resultados en consola:
+ Solución X: -2.000000 4.000000 5.000000 -5.000000 3.000000 
+ Norma del residuo: 2.220446e-16


```python
A1 = np.array([
    [0, 2, 1, 3, 4],
    [3, 1, 2, 0, 1],
    [1, 1, 1, 1, 1],
    [2, 0, 3, 1, 2],
    [4, 2, 1, 0, 1]
])

b1 = np.array([10, 11, 5, 12, 8])

x = np.linalg.solve(A1, b1)
print(x)

residu = np.subtract(A1.dot(x), b1)

norma2 = np.linalg.norm(residu)

print("Norma del residuo: ")
print(norma2)

```
